### 4.3 Eigenvalues & Eigenvectors

For the matrix:
```
A = [[4, 1],
     [2, 3]]
```

1. Compute eigenvalues and eigenvectors using `np.linalg.eig`.
2. **Verify** the eigenvector equation: `A @ v = λ * v` for each eigenpair.
3. Plot both eigenvectors as arrows, scaled by their eigenvalues, on a 2D plot.
4. **Explain** in a markdown cell (below): what does it geometrically mean for a matrix to stretch a vector?

````python
A = np.array([[4, 1], [2, 3]], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(A)  # YOUR CODE HERE

print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors (columns):\n{eigenvectors}")

# Verify Av = λv for each eigenpair
# YOUR CODE HERE
for i in range(len(eigenvalues)):
  lam=eigenvalues[i]
  v=eigenvectors[:, i]
  LHS=A @ v
  RHS=lam * v

  assert np.allclose(LHS, RHS), f"verification failed for eigenpair {i}"
print("All verified..")

# Plot
# YOUR CODE HERE
fig, ax=plt.subplots(figsize=(6, 6))
colors=['red', 'blue']
for i in range(len(eigenvalues)):
  lam=eigenvalues[i]
  v=eigenvectors[:, i]
  scaled_v=lam*v
  ax.quiver(0, 0, scaled_v[0], scaled_v[1], angles='xy', scale_units='xy', scale=1, color=colors[i], label=f'v[i+1] (lambda={lam:.1f})')
ax.set_xlim(-6, 6)
ax.set_ylim(-6, 6)
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_label('Eigenvectors scaled by Eigenvalues')
ax.axhline(0, color='black', lw=1)
ax.axvline(0, color='black', lw=1)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend()
plt.tight_layout()
plt.show()

print("4.3 passed ✓")
````

**Your Geometric Explanation (4.3):**

* **The Baseline:** Normally, when you multiply a matrix by a vector, it twists, rotates, and changes the length of that vector.
* **Eigenvectors:** These are special exceptions. When a matrix hits an eigenvector, the vector **does not rotate at all**. It stays pointed in its exact original direction.
* **Eigenvalues:** This is just a number that tells you what happens to that vector's length along its path:
    * **Stretching ($\lambda > 1$):** The matrix pulls the vector out, making it longer. If $\lambda = 5$, the vector becomes five times longer.
    * **Squishing ($0 < \lambda < 1$):** The matrix compresses the vector, pushing it down to make it shorter. If $\lambda = 0.5$, it shrinks to half its size.
    * **Flipping ($\lambda < 0$):** The vector gets stretched or squished, but it also turns around to face the exact opposite direction.
* **Summary:** Eigenvectors are the lines where a matrix **only scales things without rotating them**, and eigenvalues are the **stretch or shrink factors**.


### 4.4 SVD & Dimensionality Reduction

1. Create a random 4×3 matrix `X` (use `np.random.seed(42)`).
2. Compute its **SVD**: `U, S, Vt = np.linalg.svd(X, full_matrices=False)`.
3. **Reconstruct** `X` from the SVD and verify it matches the original.
4. Perform a **rank-1 approximation** using only the largest singular value/vector — call it `X_approx`.
5. Explain the relationship between SVD components and PCA in the markdown cell below.

````python
np.random.seed(42)
X = np.random.randn(4, 3)

U, S, Vt = np.linalg.svd(X, full_matrices=False) # YOUR CODE HERE

X_reconstructed = U @ np.diag(S) @ Vt # YOUR CODE HERE — reconstruct from U, S, Vt
X_approx        = S[0] * np.outer(U[0, :], Vt[0, :])  # YOUR CODE HERE — rank-1 approximation

print(f"U shape: {U.shape}, S shape: {S.shape}, Vt shape: {Vt.shape}")
print(f"Singular values: {S}")
print(f"Reconstruction error: {np.linalg.norm(X - X_reconstructed):.2e}")
print(f"Rank-1 approximation:\n{X_approx}")

assert np.allclose(X, X_reconstructed, atol=1e-10), "Reconstruction failed!"
print("4.4 passed ✓")
````

**SVD → PCA Connection (4.4):**

> *Explain: what are U, S, Vt in SVD, and how do they relate to principal components? Which matrix contains the directions of maximum variance?*

**SVD Components Breakdown ($X = U \cdot S \cdot V^T$):**

* **$U$ (Left Singular Vectors):** Represents your data points projected into the new coordinate space.
* **$S$ (Singular Values):** A diagonal matrix showing the strength of each component. It tells you how much variance each direction captures.
* **$V^T$ (Right Singular Vectors):** This matrix **contains the directions of maximum variance** (the principal components/loading vectors).

**The Matrix with Directions of Maximum Variance:**
* **$V^T$** (specifically, the rows of $V^T$ or the columns of $V$) holds these directions. The first row of $V^T$ points in the direction of the highest possible variance in your data.

### 5.4 Distribution Testing & Stationarity

1. Generate two samples: `s1` ~ Normal(0, 1), `s2` ~ Exponential(1), both n=200 (seed=0).
2. Run the **Kolmogorov-Smirnov test** (`stats.kstest`) comparing each against a standard normal CDF.
3. Generate a non-stationary time series (linear trend + noise). Run the **Augmented Dickey-Fuller test** (`adfuller` from `statsmodels`) and interpret the result.
4. Difference the series once and re-run ADF to confirm stationarity.

````python
from statsmodels.tsa.stattools import adfuller

np.random.seed(0)
s1 = np.random.normal(0, 1, 200)
s2 = np.random.exponential(1, 200)

# KS test — compare each against N(0,1)
ks_s1 = stats.kstest(s1, 'norm')  # YOUR CODE HERE
ks_s2 = stats.kstest(s2, 'norm')  # YOUR CODE HERE
print(f"KS test s1 (normal):      stat={ks_s1.statistic:.4f}, p={ks_s1.pvalue:.4f}")
print(f"KS test s2 (exponential): stat={ks_s2.statistic:.4f}, p={ks_s2.pvalue:.4f}")

# ADF test on non-stationary series
t = np.arange(200)
ts = 0.05 * t + np.random.normal(0, 1, 200)

adf_result = adfuller(ts)  # YOUR CODE HERE — adfuller(ts)
print(f"\nADF statistic: {adf_result[0]:.4f}")
print(f"ADF p-value:   {adf_result[1]:.4f}")
print(f"Stationary?    {adf_result[1] < 0.05}")

# Difference once and re-run
ts_diff = np.diff(ts)  # YOUR CODE HERE — np.diff
adf_diff = adfuller(ts_diff)  # YOUR CODE HERE
print(f"\nAfter differencing — p-value: {adf_diff[1]:.4f}, Stationary? {adf_diff[1] < 0.05}")
````

````python
def compute_psi(expected, actual, bins=10):
    """
    PSI = sum((actual_pct - expected_pct) * ln(actual_pct / expected_pct))
    Add epsilon=1e-10 to avoid log(0).
    """
    # YOUR CODE HERE
    epsilon=1e-10
    bin_edges=np.linspace(np.min(expected), np.max(expected), bins+1)
    bin_edges[0]=-np.inf
    bin_edges[-1]=np.inf
    expected_counts, _=np.histogram(expected, bins=bin_edges)
    actual_counts, _=np.histogram(actual, bins=bin_edges)
    expected_pct=expected_counts/len(expected)
    actual_pct=actual_counts/len(actual)
    expected_pct=np.where(expected_pct==0, epsilon, expected_pct)
    actual_pct=np.where(actual_pct==0, epsilon, actual_pct)
    psi_val=np.sum((actual_pct-expected_pct) * np.log(actual_pct/expected_pct))
    return psi_val

np.random.seed(1)
train_dist = np.random.normal(50, 10, 1000)    # training distribution
drift_dist = np.random.normal(65, 12, 1000)    # shifted production distribution

psi_value = compute_psi(train_dist, drift_dist)
print(f"PSI: {psi_value:.4f}")
print(f"Shift severity: {'Major' if psi_value > 0.2 else 'Minor' if psi_value > 0.1 else 'Stable'}")

# Plot both distributions
fig, ax = plt.subplots()
# YOUR CODE HERE
ax.hist(train_dist, bins=20, alpha=0.5, label="Training Dist (Expected)", color='gray', edgecolor='white', density=True)
ax.hist(drift_dist, bins=20, alpha=0.5, label="Production Dist (Actual)", color='crimson', edgecolor='white', density=True)
ax.legend(); ax.set_title('Distribution Shift Visualization')
plt.tight_layout(); plt.show()
````

**Concept Drift vs Covariate Drift & PSI Thresholds (5.5):**

* **Concept drift:** The relationship between your input features and target labels changes, meaning the exact same input now leads to a completely different prediction (e.g., consumer shopping habits shifting entirely due to a sudden economic change).
* **Covariate drift:** The distribution of your input features changes over time, but the underlying logic linking inputs to outputs stays the same (e.g., your app suddenly gets popular with a much younger demographic, shifting your average user age input).
* **PSI < 0.1:** No significant distribution shift; your production data remains stable and matches the training data.
* **PSI 0.1–0.2:** Minor shift; the data is beginning to drift slightly, so you should monitor the system closely.
* **PSI > 0.2:** Major shift; the data has changed significantly from the baseline training standards.
* **Retraining trigger example:** An automated pipeline that triggers a model retrain if the monthly PSI score of a core feature stays above 0.2 for two consecutive weeks.

### 6.3 Bayes' Theorem

**Scenario — Spam filter:**
- P(Spam) = 0.30 (prior)
- P("free" | Spam) = 0.80 (likelihood)
- P("free" | Not Spam) = 0.05

1. Compute P(Spam | "free") using Bayes' theorem.
2. Implement a simple `naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham)` that returns the posterior P(Spam | word).
3. In the markdown cell, map each term — **prior, likelihood, evidence, posterior** — to the spam example.

````python
p_spam            = 0.30
p_free_given_spam = 0.80
p_free_given_ham  = 0.05
p_ham             = 1 - p_spam

# P(free) — law of total probability
p_free = (p_free_given_spam * p_spam) + (p_free_given_ham * p_ham)  # YOUR CODE HERE

# Bayes
p_spam_given_free = (p_free_given_spam * p_spam)/p_free  # YOUR CODE HERE

print(f"P('free'): {p_free:.4f}")
print(f"P(Spam | 'free'): {p_spam_given_free:.4f}")

def naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham):
    # YOUR CODE HERE
    prior_ham=1.0-prior_spam
    total_evidence=(p_word_given_spam * prior_spam)+(p_word_given_ham * prior_ham)
    posterior_spam=(p_word_given_spam * prior_spam) /total_evidence
    return posterior_spam

pred = naive_bayes_predict(0.30, 0.80, 0.05)
assert abs(pred - p_spam_given_free) < 1e-9
print(f"naive_bayes_predict: {pred:.4f}")
print("6.3 passed ✓")
````

**Bayes Term Mapping (6.3):**


| Term | Formula | In this example |
|------|---------|-----------------|
| Prior | P(Spam) | **0.30** — The initial probability that an incoming email is spam before reading it. |
| Likelihood | P(word \| Spam) | **0.80** — The chance that a known spam email uses the specific word "free". |
| Evidence | P(word) | **0.275** — The total probability of the word "free" appearing in any email. |
| Posterior | P(Spam \| word) | **0.8727** — The updated probability that this email is spam because it contains "free". |


### 6.4 Central Limit Theorem

1. Define a population following an **exponential distribution** (λ=1, size=100,000).
2. Repeatedly draw samples of size `n=30` and record their means — do this 5,000 times.
3. Plot the histogram of sample means.
4. Overlay the **theoretical normal distribution** predicted by the CLT: μ = population mean, σ = population_std / sqrt(n).
5. Run a KS test to confirm the distribution of sample means is approximately normal.

````python
np.random.seed(7)
population = np.random.exponential(scale=1.0, size=100_000)

n_samples = 5000
sample_size = 30
sample_means  = np.array([np.random.choice(population, size=sample_size, replace=False).mean() for _ in range(n_samples)])  # YOUR CODE HERE — array of 5000 sample means

pop_mean = population.mean()
pop_std  = population.std()
clt_std= pop_std / np.sqrt(sample_size) # YOUR CODE HERE — predicted std of sample means

print(f"Population mean: {pop_mean:.4f}, Population std: {pop_std:.4f}")
print(f"Sample means mean: {sample_means.mean():.4f}, std: {sample_means.std():.4f}")
print(f"CLT predicted std: {clt_std:.4f}")

# Plot
fig, ax = plt.subplots()
# YOUR CODE HERE — histogram of sample means + theoretical normal overlay
ax.hist(sample_means, bins=30, edgecolor='white', color='skyblue', alpha=0.7, density=True, label='sample Means')
x_axis=np.linspace(sample_means.min(), sample_means.max(), 200)
theoratical_pdf=stats.norm.pdf(x_axis, loc=pop_mean, scale=clt_std)
ax.plot(x_axis, theoratical_pdf, color='crimson', linewidth=2, label='CLT Normal Curve')
ax.legend()

ax.set_title('CLT: Distribution of Sample Means')
plt.tight_layout(); plt.show()

# KS test
ks_result = stats.kstest(sample_means, 'norm', args=(pop_mean, clt_std))
print(f"KS test p-value: {ks_result.pvalue:.4f} → Approximately normal? {ks_result.pvalue > 0.05}")
````

**CLT Reflection (6.4):**

* **Why it matters:** The Central Limit Theorem allows us to make confident statements about a population's mean even if the underlying data is heavily skewed or non-normal, because the sample means will always form a predictable bell curve.
* **The Assumption:** This justifies the **assumption of normality** for sampling distributions, which powers foundational machine learning tools and classical tests like t-tests, ANOVAs, and confidence interval estimations.
